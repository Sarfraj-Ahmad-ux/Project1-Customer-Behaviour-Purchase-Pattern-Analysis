```python
import pandas as pd
df=pd.read_csv("C:\\Users\\Shadab\\Downloads\\customer_shopping_behavior.csv")
df.head()
```

```python
df.describe(include="all")
```

```python
df.isnull().sum()
```

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
df.columns
```

```python
#create a column age_group:
labels=['Young Adult', 'Adult', 'Middle Aged', 'Senior']
df['age_group']=pd.qcut(df['age'], q=4, labels=labels)
df[['age', 'age_group']]
```

```python
df['frequency_of_purchases'].unique()
```

```python
#creating a new column purchase_frequency_days:
mapping_dictionary={'Fortnightly':14,
                    'Weekly':7,
                    'Anually':365,
                    'Quarterly':90,
                    'Bi-Weekly':14,
                    'Monthly':30,
                    'Every 3 Months':90}
                    
df['purchase_frequency_days']=df['frequency_of_purchases'].map(mapping_dictionary)                    
```

```python
df[['purchase_frequency_days', 'frequency_of_purchases']]
```

```python
(df['promo_code_used'] == df['discount_applied']).all()
```

```python
df=df.drop('promo_code_used', axis=1)
```
